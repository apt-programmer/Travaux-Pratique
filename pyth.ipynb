# Sequential search (recherche séquentielle)

**Tableau non trié**

```python
def recherche_element1(tab, e):
    i = 0
    trouve = False
    while i < len(tab) and not trouve:
        if tab[i] == e:
            trouve = True
        else:
            i += 1
    return trouve

# Example usage
if __name__ == "__main__":
  tab = [9, 5, 11, 20, 3]
  e = 25
  print(recherche_element1(tab, e))  # Output: True

  tab = ["apple", "banana", "cherry"]
  e = "banana"
  print(recherche_element1(tab, e))  # Output: True
```

**Optimisation sur tableau trié**

```python
def recherche_element2(tab, e):
    i = 0
    trouve = False
    possible = True

    while i < len(tab) and possible and not trouve:
        if tab[i] == e:
            trouve = True
        elif tab[i] > e:
            possible = False  # e cannot appear later due to sorted order
        else:
            i += 1

    return trouve

# Example usage
if __name__ == "__main__":
  tab = ["apple", "banana", "cherry", "date", "fig"]
  e = "carotte"
  print(recherche_element2(tab, e))  # Output: False
```

**Variante  : Retourner l'indice de la valeur recherchée**

```python
def sequential_search(arr, target):
    for index, value in enumerate(arr):
        if value == target:
            return index  # Return the index of the found item
    return -1  # Return -1 if the target is not found

# Example usage
if __name__ == "__main__":
  data = [10, 25, 37, 49, 55]
  target = 49

  result = sequential_search(data, target)
  print(f"Target {target} found at index: {result}" if result != -1 else "Target not found.")
```

# Binary search (recherche Dichotomique)

**Iterative Binary Search**

```python
def binary_search(arr, e):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == e:
            return mid  # Or return True if only presence is needed
        elif arr[mid] < e:
            left = mid + 1
        else:
            right = mid - 1

    return -1  # Or return False

# Example usage
if __name__ == "__main__":
  data = [2, 5, 7, 8, 11, 12]
  target = 11

  result = binary_search(data, target)
  print(f"Target {target} found at index: {result}" if result != -1 else "Target not found.")
```

**Recursive Binary Search**

```python
def binary_search_recursive(arr, target, left, right):
    if left > right:
        return -1  # Base case: not found

    mid = (left + right) // 2

    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search_recursive(arr, target, mid + 1, right)
    else:
        return binary_search_recursive(arr, target, left, mid - 1)

# Example usage
if __name__ == "__main__":
  tab = [3, 8, 15, 23, 42, 57, 89]
  target = 57
  result = binary_search_recursive(tab, target, 0, len(tab) - 1)
  print(f"Target {target} found at index: {result}" if result != -1 else "target not found")  # Output: 3
```

**Interpolation Binary Search**

```python
def interpolation_search(tab, e):
    i, j = 0, len(tab) - 1
    while i <= j and e >= tab[i] and e <= tab[j]:
        if tab[i] == tab[j]:
            return i if tab[i] == e else -1
        a = (tab[j] - tab[i]) / (j - i)
        b = tab[i] - a * i
        k = int((e - b) / a)
        if k < i or k > j:
            break
        if tab[k] == e:
            return k
        elif tab[k] < e:
            i = k + 1
        else:
            j = k - 1
    return -1

# Example usage
if __name__ == "__main__":
    tab = [10, 20, 30, 40, 50, 60, 70, 80, 90]
    e = 90
    print(f"Element {e} found at index: {interpolation_search(tab, e)}")
```

# Verify sorting

```python
def test_trie(tab):
    for i in range(len(tab) - 1):
        if tab[i] > tab[i + 1]:
            return False # Quit immediately if unsorted
    return True

# Example usage
if __name__ == "__main__":
    arr = [1, 2, 3, 4, 6, 5]
    print("Tableau trié:", test_trie(arr)) # Output: False

```

# Bubble sort

```python
def tri_bulle(tab):
    n = len(tab)
    # i va de n-2 à 1 (inclus) par pas de -1
    for i in range(n - 2, 0, -1):
        # j va de 0 à i (inclus)
        for j in range(i + 1):
            if tab[j] > tab[j + 1]:
                # échange des éléments
                tab[j], tab[j + 1] = tab[j + 1], tab[j]

# Exemple d'utilisation
if __name__ == "__main__":
    arr = [13, 5, 3, -8, 4, -5, 2, 7]
    #arr = "barthqsmfhdl"
    tri_bulle(arr)
    print("Tableau trié :", arr)
```

**Opimized Bubble sort**

```python
def tri_bulle(tab):
    n = len(tab)
    for i in range(n - 2, 0, -1):
        swapped = False
        for j in range(i + 1):
            if tab[j] > tab[j + 1]:
                tab[j], tab[j + 1] = tab[j + 1], tab[j]
                swapped = True
        if not swapped:
            break  # Exit early if the array is already sorted

# Example usage
if __name__ == "__main__":
    arr = [13, 5, 3, -8, 4, 5, -2, 7]
    tri_bulle(arr)
    print("Sorted array:", arr)
```

**Recursive Bubble sort**

```python
def recursive_bubble_sort(arr, n):
    # Base case: if the array size is 1, it is already sorted.
    if n <= 1:
        return

    # Perform one pass to verify if the arr is already sorted.
    swapped = False
    for i in range(n - 1):
        if arr[i] > arr[i + 1]:
            arr[i], arr[i + 1] = arr[i + 1], arr[i]
            swapped = True

    # If no swaps occurred, the array is already sorted.
    if not swapped:
        return

    # Recursively sort the first n-1 elements.
    recursive_bubble_sort(arr, n - 1)
    #return arr

# Example usage
if __name__ == "__main__":
    arr = [64, 34, 25, 12, 22, 11, -90]
    recursive_bubble_sort(arr, len(arr))
    print("Sorted array:", arr)
```

# Selection sort

```python
def tri_selection(tab):
    for i in range(len(tab) - 1, 0, -1):  # Iterate from len(tab)-1 down to 1
        pg = 0  # Initialize the index of the largest element
        for j in range(i + 1):  # Iterate through the unsorted portion
            if tab[j] > tab[pg]:
                pg = j  # Update pg if a larger element is found
        # Swap the largest element with the element at index i
        tab[pg], tab[i] = tab[i], tab[pg]

# Example usage
if __name__ == "__main__":
    arr = [25, 64, 12, 22, 11, 11, -13]
    tri_selection(arr)
    print("Sorted array:", arr)
```

**Recursive Selection sort**

```python
def tri_selection_recursive(tab, n):
    # Base case: if the unsorted portion has 0 or 1 element, it is sorted.
    if n <= 1:
        return

    # Find the index of the largest element in the unsorted portion (0 to n-1).
    pg = 0
    for j in range(n):
        if tab[j] > tab[pg]:
            pg = j

    # Swap the largest element with the last element in the unsorted portion.
    tab[pg], tab[n - 1] = tab[n - 1], tab[pg]

    # Recursively sort the unsorted portion (from 0 to n-2).
    tri_selection_recursive(tab, n - 1)

# Example usage
if __name__ == "__main__":
    arr = [-64, 25, 12, 22, 22, 11]
    tri_selection_recursive(arr,len(arr))
    print("Sorted array:", arr)
```

# Insertion sort

```python
def tri_insertion(tab):
    for i in range(1, len(tab)):
        val = tab[i] # mémoriser la valeur à insérer.
        j = i # reculer dans la partie triée pour trouver la bonne position.
        while j > 0 and tab[j - 1] > val: # décaler les éléments plus grands que val vers la droite.
            tab[j] = tab[j - 1]
            j -= 1
        tab[j] = val # insérer val à la bonne position.

# Exemple d'utilisation
if __name__ == "__main__":
    arr = [7, 3, 5, 2, 8, -1]
    tri_insertion(arr)
    print("Tableau trié :", arr)
```

**Recursive Insertion Sort**

```python
def tri_insertion_recursive(tab, n):
    if n <= 1:
        return

    tri_insertion_recursive(tab, n - 1)

    val = tab[n - 1]
    j = n - 1
    while j > 0 and tab[j - 1] > val:
        tab[j] = tab[j - 1]
        j -= 1
    tab[j] = val

# Example usage
if __name__ == "__main__":
    arr = [7, 3, 5, 2, 8, -1, -93]
    tri_insertion_recursive(arr, len(arr))
    print("Tableau trié :", arr)
```

# Merge sort (tri par fusion)

```python
def fusion(gauche, droite):
    resultat = []
    i = j = 0

    # Comparaison et fusion élément par élément
    while i < len(gauche) and j < len(droite):
        if gauche[i] <= droite[j]:
            resultat.append(gauche[i])
            i += 1
        else:
            resultat.append(droite[j])
            j += 1

    # Ajouter les éléments restants (s'il y en a)
    resultat.extend(gauche[i:])
    resultat.extend(droite[j:])
    return resultat

def merge_sort(tab):
    # Cas de base : un tableau de taille 0 ou 1 est déjà trié
    if len(tab) <= 1:
        return tab

    # Étape 1 : diviser le tableau en deux moitiés
    milieu = len(tab) // 2
    gauche = merge_sort(tab[:milieu])
    droite = merge_sort(tab[milieu:])

    # Étape 2 : fusionner les deux moitiés triées
    return fusion(gauche, droite)

# Exemple d'utilisation
if __name__ == "__main__":
    tableau = [38, 27, 43, 3, 9, 82, 10, 10, -1]
    trié = merge_sort(tableau)
    print("Tableau trié :", trié)
```

# Quick sort (tri par rapide)

```python
def quick_sort(arr):
    # Base case: arrays with 0 or 1 element are already sorted
    if len(arr) <= 1:
        return arr

    # Choose pivot (here: middle element or first or last element)
    pivot = arr[len(arr) // 2]

    # Partition into three lists
    less  = [x for x in arr if x < pivot]
    equal = [x for x in arr if x == pivot]
    greater = [x for x in arr if x > pivot]

    # Recursively sort partitions and concatenate
    return quick_sort(less) + equal +  quick_sort(greater)

# Example usage
if __name__ == "__main__":
    sample = [33, 10, 55, 71, 29,19,3, 3, 18, 42]
    sorted_sample = quick_sort(sample)
    print("Sorted array:", sorted_sample)
```

```python
def quick_sort(arr):
    # Base case: arrays with 0 or 1 element are already sorted
    if len(arr) <= 1:
        return arr

    # Choose pivot as the last element
    pivot = arr[-1]

    # Partition into three lists, excluding the pivot
    less = [x for x in arr if x < pivot]
    equal = [x for x in arr if x == pivot]
    greater = [x for x in arr if x > pivot]

    # Recursively sort partitions and concatenate, including the pivot
    return quick_sort(less) + equal + quick_sort(greater)

# Example usage
if __name__ == "__main__":
    sample = [33, 10, 55, 71, 29, 19, 29, 3, 18, 42, 0 , -1]
    sorted_sample = quick_sort(sample)
    print("Sorted array:", sorted_sample)
```
